The main objective is to identify crop production patterns, compare agricultural performance across regions, and understand seasonal and yearly production trends.

## Project Objectives

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path

# Locate the project folder
project_folder = Path.cwd().parent.parent

# Find the dataset
csv_files = list((project_folder / "Dataset").glob("*.csv"))

# Load the dataset
df = pd.read_csv(csv_files[0])

print("Dataset loaded successfully!")
print("Total rows:", df.shape[0])
print("Total columns:", df.shape[1])
```

```python
# Dataset Overview

print("Dataset Shape:", df.shape)

print("\nColumn Names:")
print(df.columns.tolist())

print("\nDataset Information:")
df.info()


```

```python
#3. CHECK DATA TYPES
# ------------------------------------------------------------

print("\n--- DATA TYPES ---")
print(df.dtypes)


# ------------------------------------------------------------
# 4. MISSING VALUES
# ------------------------------------------------------------

print("\n--- MISSING VALUES ---")

missing_values = df.isnull().sum()

missing_report = pd.DataFrame({
    "Missing Values": missing_values,
    "Missing Percentage": (
        df.isnull().sum() / len(df) * 100
    ).round(2)
})

display(missing_report)

print(
    "\nTotal missing Production values:",
    df["Production"].isnull().sum()
)


# ------------------------------------------------------------
# 5. DUPLICATE RECORDS
# ------------------------------------------------------------

print("\n--- DUPLICATE RECORDS ---")

duplicate_count = df.duplicated().sum()

print("Total duplicate rows:", duplicate_count)
```
